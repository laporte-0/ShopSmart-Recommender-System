# Reciprocal rank fusion and recommendation demonstration

Each component contributes weight/(60+rank) over its top 100 unseen candidates.
Absent scores contribute zero; popularity fills candidates with no contribution.
Weights use validation only. The hybrid is compared against each component and
is not assumed to win. All results are development estimates, not independent tests.


```python
from pathlib import Path
import sys
import pandas as pd
import joblib
ROOT = Path.cwd() if (Path.cwd() / 'reports').exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT))
results = pd.read_csv(ROOT / 'reports/results/validation_all_runs.csv')
bundle = joblib.load(ROOT / 'artifacts/recommenders.joblib')
```

```python
display(results[results.family.eq('Hybrid')][['model','hyperparameters','Hit Rate@10','NDCG@10','Coverage','Diversity']])
display(pd.read_csv(ROOT / 'reports/results/validation_selected.csv'))
```

```python
data=bundle['data']
user=data.users[0]
print('User:', user)
display(data.train[data.train.user_id.eq(user)].merge(data.products,on='parent_asin')[['title','rating']].head())
for name in [bundle['winner'], 'Popularity']:
    recs=bundle['models'][name].recommend(user)
    frame=pd.DataFrame(recs,columns=['parent_asin','ranking_score']).merge(data.products,on='parent_asin')
    print(name, 'ranking scores, not calibrated ratings')
    display(frame[['title','categories','ranking_score']])
```
